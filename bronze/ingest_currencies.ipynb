```python
import requests

url = "https://olinda.bcb.gov.br/olinda/servico/PTAX/versao/v1/odata/Moedas?top=100&format=json&$select=simbolo,nomeFormatado"
```

```python
response = requests.get(url)
data = response.json()['value']
df = spark.createDataFrame(data)
df = df.selectExpr(
    "nomeFormatado AS dc_currency_name",
    "simbolo AS dc_currency_code"
)

df.write.mode('overwrite').option("overwriteSchema", "true").saveAsTable('currency.03_gold_currency.dim_currencies')
display(df)
```
