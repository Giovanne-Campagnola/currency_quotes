```python
import requests

url = "https://olinda.bcb.gov.br/olinda/servico/PTAX/versao/v1/odata/Moedas?top=100&format=json&$select=simbolo,nomeFormatado"
```

```python
response = requests.get(url)
dados = response.json()['value']
df = spark.createDataFrame(dados)
df = df.selectExpr(
    "nomeFormatado AS dc_moeda_nome",
    "simbolo AS dc_moeda_abrev"
)

df.write.mode('overwrite').option("overwriteSchema", "true").saveAsTable('workspace.03_gold_currency.dim_moedas')
display(df)
```
